```python
labels = [1,1,0,0,0,0,1,0,0,0]
labels = np.array(labels)
print(labels)
unique_lbls = np.unique(labels)
print(unique_lbls)
```

```python
entropy = 0
for lbl in unique_lbls:
    percentage = len(np.extract(labels == lbl, labels))/len(labels)
    print(percentage)
    entropy -= percentage * np.log2(percentage)

print(entropy)
```
